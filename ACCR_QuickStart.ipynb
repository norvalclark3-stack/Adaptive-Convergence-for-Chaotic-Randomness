```python
import numpy as np

# ------------------------------------------------------------
# Minimal ACCR Operators (Quick-Start Version)
# ------------------------------------------------------------

def D(x):
    """Directional change."""
    return np.diff(x)

def DD(x):
    """Acceleration of change."""
    return np.diff(D(x))

def G(x):
    """Instability gradient."""
    return np.abs(DD(x))

def F(x):
    """Collapse funnel indicator (normalized gradient acceleration)."""
    g = G(x)
    return g / (np.max(g) + 1e-9)

def EWS(x):
    """Early-warning signal."""
    return G(x) * F(x)
```
